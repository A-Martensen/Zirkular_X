# Loading Modules/Packages


```python
from pathlib import Path 
from rdflib import Graph, RDF, RDFS, OWL
```

```python
base =Path(".")
g = Graph()

for ttl in sorted(base.glob("DPPO/*.ttl")) + sorted(base.glob("RUB/*.ttl")):
    g.parse(ttl, format="turtle")
    print(f"{ttl}: total triples now {len(g)}")

classes = set(g.subjects(RDF.type, OWL.Class))
obj_props = set(g.subjects(RDF.type, OWL.ObjectProperty))
data_props = set(g.subjects(RDF.type, OWL.DatatypeProperty))
print(f"\nClasses: {len(classes)}, ObjectProperties: {len(obj_props)}, DatatypeProperties: {len(data_props)}")
```

```python
from pathlib import Path
from rdflib import Graph

g = Graph()

for ttl in sorted(Path(".").glob("**/*.ttl")):
    try:
        gi = Graph().parse(ttl, format="turtle")
    except (SyntaxError, OSError, UnicodeDecodeError) as e:
        print(f"ERROR  {ttl}\n       {e}")
        continue
    print(f"OK     {ttl}  ({len(gi)} triples)")
    g += gi

print(f"\nCombined: {len(g)} triples")
```
